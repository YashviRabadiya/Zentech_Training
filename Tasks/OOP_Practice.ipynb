# OOP

### Class and Objects

```python
class Person:
	def __init__(self,name,age):
		self.name = name
		self.age = age
	def greet(self):
		print(f"Hello, my name is {self.name}")

# Create an object
p1 = Person('Yashvi',20)
# Call the greet method
p1.greet()
```

### magic methods also called dunder method (double underscore)

```python
# __init__ and __str__

class Person:
  def __init__(self, name):
    self.name = name

  def __str__(self):
    return f"Person: {self.name}"

p1 = Person("Yashvi")
print(p1)
```
